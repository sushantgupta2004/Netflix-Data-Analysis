# 🎬 Netflix Data Analysis

## Objective
Analyze Netflix movies and TV shows to identify patterns in content type, release trends, countries, ratings, genres and duration.

**Tools:** Python, Pandas, NumPy, Matplotlib

> Dataset source: TidyTuesday / public Netflix titles dataset.


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_URL = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2021/2021-04-20/netflix_titles.csv"
LOCAL_PATH = "../data/netflix_titles.csv"

if os.path.exists(LOCAL_PATH):
    df = pd.read_csv(LOCAL_PATH)
else:
    df = pd.read_csv(DATA_URL)

df.head()
```

## 1. Dataset Overview

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
df.info()
```

```python
df.describe(include="all").T
```

## 2. Data Cleaning

```python
df = df.drop_duplicates()

df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
df["added_year"] = df["date_added"].dt.year
df["added_month"] = df["date_added"].dt.month_name()

df["movie_minutes"] = pd.to_numeric(
    df["duration"].str.extract(r"(\d+)")[0], errors="coerce"
)

df["seasons"] = pd.to_numeric(
    df["duration"].str.extract(r"(\d+)")[0], errors="coerce"
)

print("Duplicate rows removed.")
print("Current shape:", df.shape)
```

## 3. Missing Values

```python
missing = (
    df.isna().sum()
      .sort_values(ascending=False)
      .to_frame("missing_values")
)
missing["missing_percent"] = (missing["missing_values"] / len(df) * 100).round(2)
missing
```

## 4. Movies vs TV Shows

```python
content_type = df["type"].value_counts()
content_type
```

```python
plt.figure(figsize=(7,5))
content_type.plot(kind="bar")
plt.title("Netflix Content: Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 5. Titles Added by Year

```python
titles_by_year = df["added_year"].value_counts().sort_index()
titles_by_year.tail(15)
```

```python
plt.figure(figsize=(10,5))
titles_by_year.plot(kind="line", marker="o")
plt.title("Netflix Titles Added by Year")
plt.xlabel("Year")
plt.ylabel("Titles Added")
plt.tight_layout()
plt.show()
```

## 6. Top Countries

```python
top_countries = (
    df["country"].dropna()
      .str.split(", ")
      .explode()
      .value_counts()
      .head(10)
)
top_countries
```

```python
plt.figure(figsize=(9,5))
top_countries.sort_values().plot(kind="barh")
plt.title("Top 10 Countries by Netflix Titles")
plt.xlabel("Titles")
plt.tight_layout()
plt.show()
```

## 7. Ratings

```python
ratings = df["rating"].value_counts().head(10)
ratings
```

```python
plt.figure(figsize=(9,5))
ratings.sort_values().plot(kind="barh")
plt.title("Most Common Netflix Ratings")
plt.xlabel("Titles")
plt.tight_layout()
plt.show()
```

## 8. Top Genres

```python
top_genres = (
    df["listed_in"].dropna()
      .str.split(", ")
      .explode()
      .value_counts()
      .head(10)
)
top_genres
```

```python
plt.figure(figsize=(9,5))
top_genres.sort_values().plot(kind="barh")
plt.title("Top 10 Netflix Genres")
plt.xlabel("Titles")
plt.tight_layout()
plt.show()
```

## 9. Movie Duration

```python
movie_duration = df.loc[df["type"] == "Movie", "movie_minutes"].dropna()
movie_duration.describe()
```

```python
plt.figure(figsize=(8,5))
plt.hist(movie_duration, bins=30)
plt.title("Movie Duration Distribution")
plt.xlabel("Minutes")
plt.ylabel("Number of Movies")
plt.tight_layout()
plt.show()
```

## 10. TV Show Seasons

```python
tv_seasons = df.loc[df["type"] == "TV Show", "seasons"].dropna()
tv_seasons.value_counts().sort_index().head(10)
```

```python
plt.figure(figsize=(8,5))
tv_seasons.value_counts().sort_index().head(10).plot(kind="bar")
plt.title("TV Shows by Number of Seasons")
plt.xlabel("Seasons")
plt.ylabel("Shows")
plt.tight_layout()
plt.show()
```

## 11. Business-Oriented Insights

```python
print("Total titles:", len(df))
print("\nContent mix:")
print(df["type"].value_counts())

print("\nTop countries:")
print(top_countries.head(5))

print("\nTop genres:")
print(top_genres.head(5))

print("\nMost common ratings:")
print(ratings.head(5))
```

### Key Takeaways to Discuss in an Interview

1. Netflix's catalogue contains both movies and TV shows, allowing analysis of its content mix.
2. Content additions can be examined over time to identify periods of catalogue expansion.
3. Country-level analysis helps understand the geographic distribution of content.
4. Rating analysis can be used to understand the maturity profile of the catalogue.
5. Genre analysis highlights the types of content represented most frequently.
6. Duration and season analysis helps compare movie length with TV-series structure.

These are descriptive findings from the dataset, not claims about Netflix's current catalogue or strategy.
